# Exploratory Data Analysis (EDA)

## Objective
Explore patterns and relationships in the IBM HR Attrition dataset to identify key drivers of employee attrition.

## Focus Areas
- Attrition distribution
- Demographic and job-related factors
- Compensation and experience
- Early indicators of employee turnover

## Outcome
Actionable insights to support HR decision-making.
01_exploratory_data_analysis.ipynb

### Import Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style='whitegrid')
plt.rcParams["figure.figsize"] = (10, 6)
```

### Load Clean Dataset

```python
df = pd.read_csv("../01_dataset/raw/IBM_HR_Analytics_Employee_Attrition_Performance.csv")
```

### EDA Analysis

### Attrition Distribution

```python
df['Attrition'].value_counts()
```

```python
sns.countplot(x='Attrition', data=df)
plt.title("Employee Attrition Distribution")
plt.show()
```

### Insight
- Majority of employees stayed with the company
- Attrition class imbalance exists
- Predictive models will require imbalance handling

### Attrition by Gender

```python
sns.countplot(x='Gender', hue='Attrition', data=df)
plt.title("Attrition by Gender")
plt.show()
```

### Insights
- Attrition occurs both genders
- Slightly higher attrition observed in male employees

#### Feature candidate: Gender

### Attrition by Department

```python
sns.countplot(x='Department', hue='Attrition', data=df)
plt.title("Attrition by Department")
plt.xticks(rotation=20)
plt.show()
```

### Attrition by Job Role

```python
plt.figure(figsize=(10,5))
sns.countplot(
    data=df,
    y='JobRole',
    hue='Attrition'
)
plt.title('Attrition by Job Role')
plt.xlabel('Employee Count')
plt.ylabel('Job Role')
plt.show()
```

- Certain job roles show consistently higher attrition.
- Operational and frontline roles appear more vulnerable to turnover.
- These roles should be prioritized for retention strategies.

### Feture relevance: JobRole (Categorical)

### Age vs Attrition

```python
plt.figure(figsize=(8,5))
sns.boxenplot(
    data=df,
    x='Attrition',
    y='Age'
)
plt.title('Age Distribution by Attrition')
plt.show()
```

- Employees who leave tend to be younger.
- Early-career attrition is a key concern for the company.

#### Feature: Age

### Monthly Income vs Attrition

```python
plt.figure(figsize=(8,5))
sns.boxplot(
    data=df,
    x='Attrition',
    y='MonthlyIncome'
)
plt.title('Monthly Income vs Attrition')
plt.show()
```

- Employees with lower income show higher attrition.
- Compensation is a strong retention factor.

#### Feature: MonthlyIncome

### Years at Company vs Attrition

```python
sns.boxenplot(
    data = df,
    x = 'Attrition',
    y = 'YearsAtCompany'
)
plt.title('Years at Company vs Attrition')
plt.show()
```

- Employees tend to leave within early years of employment.
- Long-term employees show higher retention.

#### Feature: YearsAtCompany

### Overtime vs Attrition

```python
sns.countplot(
    data = df,
    x = 'OverTime',
    hue = 'Attrition'
)
plt.title('Overtime vs Attrition')
plt.show()
```

- Employees working overtime have significantly higher attrition.
- Work-life balance is a major attrition driver.

#### Feature: OverTime

### Correlation Heatmap

```python
plt.figure(figsize=(12,8))
sns.heatmap(
    df.corr(numeric_only=True),
    cmap='coolwarm',
    annot=False
)
plt.title('Correlation Heatmap')
plt.show()
```

- Income, age, and total working years show meaningful correlations.
- HIghly correlated variables will be handled during feature selection.

## Key EDA Findings

Based on exploratory analysis, the following factors strongly influence employee attrition:

- Job Role
- Department
- Age
- Monthly Income
- Years at Company
- Over Time
- Work Life Balance
- Job Satisfaction

These variables will be used as input features for predictive modeling in the next phase.
