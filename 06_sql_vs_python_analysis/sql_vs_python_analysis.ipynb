# SQL vs Python Analysis – Employee Attrition

## Objective
The purpose of this notebook is to validate employee attrition insights
by replicating SQL-style analytical logic using Python (Pandas).

This ensures analytical consistency and demonstrates cross-tool proficiency.


Import Libraries

```python
import pandas as pd
```

Load Dataset

```python
df = pd.read_csv("../01_dataset/raw/IBM_HR_Analytics_Employee_Attrition_Performance.csv")
```

## Dataset Overview

The dataset contains 1,470 employee records with 35 features,

## Q1: Attrition by Department

**Business Question:**  
Which departments experience the highest employee attrition?

SQL Logic:
- GROUP BY Department
- COUNT total employees
- COUNT employees where Attrition = 'Yes'
- Calculate Attrition Rate


Python Execution

```python
attrition_by_dept = (
    df.groupby("Department")
      .agg(
          Total_Employees=("Attrition", "count"),
          Attrition_Count=("Attrition", lambda x: (x == "Yes").sum())
      )
)

attrition_by_dept["Attrition_Rate"] = (
    attrition_by_dept["Attrition_Count"] /
    attrition_by_dept["Total_Employees"] * 100
).round(2)

attrition_by_dept.sort_values("Attrition_Rate", ascending=False)
```

**Insight:**  
Sales and Human Resources departments show the highest attrition rates,
indicating higher pressure, target-driven roles, and possible workload imbalance.


## Q2: Attrition by Job Role

```python
jobrole_attrition = (
    df.groupby("JobRole")["Attrition"]
      .value_counts()
      .unstack()
      .fillna(0)
      .sort_values(by="Yes", ascending=False)
)

jobrole_attrition
```

**Insight:**  
Laboratory Technicians and Sales roles show consistently higher attrition,
suggesting shift stress, performance pressure, and limited growth opportunities.


## Q3: Attrition vs Overtime

```python
overtime_attrition = (
    df.groupby("OverTime")['Attrition']
    .value_counts()
    .unstack()
    .fillna(0)
)

overtime_attrition
```

**Insight:**  
Employees working overtime show significantly higher attrition,
highlighting work-life imbalance as a major exit driver.


## Q4: Monthly Income vs Attrition

```python
df.groupby("Attrition")['MonthlyIncome'].mean().round(2)
```

**Insight:**  
Employees who left the organization earn substantially less on average,
confirming compensation as a strong attrition driver.

The results obtained using Python (Pandas) fully align with SQL-based
