## Step 11A: Feature Engineering

### Objective
The goal of this step is to transform the cleaned HR dataset into a machine-learning-ready format.
This involves encoding categorical variables, separating features and target variables,
and ensuring that all inputs are numerical and interpretable for predictive models.

Feature engineering bridges the gap between exploratory analysis and predictive modeling.


### Load Libraries & Dataset

```python
import pandas as pd 
import numpy as np
```

```python
df = pd.read_csv("../01_dataset/raw/IBM_HR_Analytics_Employee_Attrition_Performance.csv")
df.head()
```

### Target Variable Preparation

🎯 Business Logic

Attrition is a binary classification problem.

```python
df['Attrition'].value_counts()
```

```python
df['Attrition'] = df['Attrition'].map({'Yes': 1, 'No': 0})
```

Attrition is encoded as a binary target variable:
- Yes → 1 (Employee left)
- No → 0 (Employee stayed)

This encoding enables classification algorithms to learn attrition patterns.


### Drop Non-Predictive Identifier Columns

These columns do not help prediction and may cause noise.

```python
non_predictive_cols = ['EmployeeNumber', 'EmployeeCount', 'StandardHours']
df.drop(columns=non_predictive_cols, inplace=True)
```

Identifier and constant columns were removed because they do not contribute
to predictive performance and may introduce noise into the model.


### Separate Categorical & Numerical Features

```python
categorical_cols = df.select_dtypes(include='object').columns.tolist()
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()

categorical_cols, numerical_cols
```

Features were grouped into categorical and numerical variables
to apply appropriate encoding techniques.

### Encode Categorical Variables (One-Hot Encoding)

💡 Why One-Hot?

Avoids ordinal assumptions

Works well for tree & linear models

```python
df_encoded = pd.get_dummies(df, columns=categorical_cols, drop_first=True)
```

Categorical variables were converted into numerical format using one-hot encoding.
The first category was dropped to avoid multicollinearity (dummy variable trap).


### Split Features & Target

```python
X = df_encoded.drop('Attrition', axis=1)
y = df_encoded['Attrition']

X.shape, y.shape
```

The dataset was split into:
- X: Input features
- y: Target variable (Attrition)

This separation prepares the data for model training and evaluation.


### Final Dataset Check

```python
X.head()
```

```python
y.value_counts(normalize=True)
```

The final dataset is fully numerical, model-ready, and suitable
for both interpretable and advanced machine learning algorithms.
