## Model Evaluation & Comparison

The objective of this step is to evaluate and compare all trained machine
learning models using appropriate performance metrics.

Since employee attrition is a class-imbalanced problem, the focus is placed on:
- Recall for Attrition (Yes)
- Precision–Recall trade-off
- F1-score
- Confusion Matrix analysis

The goal is to select a model that best supports proactive HR decision-making
by correctly identifying employees at risk of leaving.


### Models Evaluated
- Logistic Regression (Baseline & Interpretable)
- Decision Tree Classifier
- Random Forest Classifier


```python
import pandas as pd

model_comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "Random Forest"
    ],
    "Accuracy": [
        0.75,
        0.77,
        0.84
    ],
    "Recall (Attrition = Yes)": [
        0.62,
        0.53,
        0.09
    ],
    "Precision (Attrition = Yes)": [
        0.35,
        0.36,
        0.44
    ],
    "F1 Score (Attrition = Yes)": [
        0.44,
        0.43,
        0.14
    ]
})

model_comparison
```

### Visual Comparison

```python
model_comparison.set_index("Model")[[
    "Recall (Attrition = Yes)",
    "F1 Score (Attrition = Yes)"
]].plot(kind="bar", figsize=(10,5), title="Model Comparison on Attrition Detection")
```

### Confusion Matrix Interpretation

- False Negatives are the most costly errors in attrition prediction
- Missing an employee who is about to leave prevents proactive HR intervention
- Models with higher recall are preferred even at the cost of accuracy


## Model Selection Decision

Although Random Forest achieved the highest accuracy, it performed poorly in
identifying actual attrition cases.

Logistic Regression demonstrated the best balance between:
- Interpretability
- Recall performance
- Business usability

Therefore, Logistic Regression is selected as the preferred model for
employee attrition risk prediction.


| Criteria | Best Model |
|-------|-----------|
| Overall Accuracy | Random Forest |
| Attrition Detection | Logistic Regression |
| Interpretability | Logistic Regression |
| HR Business Usability | Logistic Regression |
